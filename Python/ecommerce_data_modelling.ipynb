```python
import pandas as pd
events= pd.read_csv("/content/events.csv")
orders= pd.read_csv("/content/orders.csv")
order_items = pd.read_csv("/content/order_items.csv")
products= pd.read_csv("/content/products.csv")
reviews= pd.read_csv("/content/reviews.csv")
users= pd.read_csv("/content/users.csv")

datasets= {

"Events": events,
"Orders": orders,
"Order items": order_items,
"Products": products,
"Reviews" : reviews,
"Users" : users

}
```

```python
for name, df in datasets.items():
    print("="*60)
    print(name)
    print("="*60)
    print("Shape:", df.shape)
    print("\nColumns:")
    print(df.columns.tolist())
    print("\nInfo:")
    print(df.info())
    print("\nFirst Five Rows:")
    print(df.head())
    print("\nMissing Values:")
    print(df.isnull().sum())
    print("\n")
```

```python
total_revenue= orders['total_amount'].sum()
total_orders = orders['order_id'].nunique()
total_customers= users['user_id'].nunique()
aov= total_revenue/total_orders
total_products_sold= order_items['quantity'].sum()
avg_reviews = reviews['rating'].mean()

print("Total Revenue:",round(total_revenue,2))
print("Total Orders:",round(total_orders,2))
print("Total Customers:",round(total_customers,2))
print("Average order value:",round(aov,2))
print("Total products sold :",round(total_products_sold,2))
print("Average reviews:",round(avg_reviews,2))

```

```python
orders['year'] = orders['order_date'].dt.year
orders['month_num'] = orders['order_date'].dt.month
orders['month_name'] = orders['order_date'].dt.month_name()
```

```python
monthly_revenue = (
    orders.groupby(['year', 'month_num', 'month_name'])['total_amount']
    .sum()
    .reset_index()
    .sort_values(['year', 'month_num'])
)

monthly_revenue
```
